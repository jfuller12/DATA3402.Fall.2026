*Exercise 4:* Write a function that takes an `n×n` matrix representing a tic‑tac‑toe game and returns one of the following values:

- `-1` if the game is **incomplete** (still empty spaces and no winner)
- `0` if the game is a **draw**
- `1` if **player 1** (`"X"`) has won
- `2` if **player 2** (`"O"`) has won

Here are some example inputs you can use to test your code:


````python
def check_game(board):
    n = len(board)

    # check rows
    for i in range(n):
        if board[i][0] != 0:
            player = board[i][0]
            won = True
            for j in range(n):
                if board[i][j] != player:
                    won = False
            if won:
                return player

    # check columns
    for j in range(n):
        if board[0][j] != 0:
            player = board[0][j]
            won = True
            for i in range(n):
                if board[i][j] != player:
                    won = False
            if won:
                return player

    # check main diagonal
    if board[0][0] != 0:
        player = board[0][0]
        won = True
        for i in range(n):
            if board[i][i] != player:
                won = False
        if won:
            return player

    # check other diagonal
    if board[0][n - 1] != 0:
        player = board[0][n - 1]
        won = True
        for i in range(n):
            if board[i][n - 1 - i] != player:
                won = False
        if won:
            return player

    # check for empty spaces
    for i in range(n):
        for j in range(n):
            if board[i][j] == 0:
                return -1

    # if there are no empty spaces, it is a draw
    return 0
````

*Exercise 12:* Develop a version of the game where one player is the computer. This time, write a computer player using exhaustive search with a max depth parameter, similar to lecture.

````python
def create_board(n):
    board = []
    for i in range(n):
        row = []
        for j in range(n):
            row.append(0)
        board.append(row)
    return board

def draw_board(board):
    n = len(board)

    # print column labels
    print("    ", end="")

    for j in range(n):
        print(chr(65 + j) + "   ", end="")
        
    print()

    for i in range(n):
        print("  " + "--- " * n)

        # print row number
        print(str(i + 1) + " ", end="")
        for j in range(n):
            if board[i][j] == 0:
                print("|   ", end="")
            elif board[i][j] == 1:
                print("| X ", end="")
            elif board[i][j] == 2:
                print("| O ", end="")
        print("|")

    print("  " + "--- " * n)

def check_game(board):
    n = len(board)

    # check rows
    for i in range(n):
        if board[i][0] != 0:
            player = board[i][0]
            won = True
            for j in range(n):
                if board[i][j] != player:
                    won = False
            if won:
                return player

    # check columns
    for j in range(n):
        if board[0][j] != 0:
            player = board[0][j]
            won = True
            for i in range(n):
                if board[i][j] != player:
                    won = False
            if won:
                return player

    # check main diagonal
    if board[0][0] != 0:
        player = board[0][0]
        won = True
        for i in range(n):
            if board[i][i] != player:
                won = False
        if won:
            return player

    # check other diagonal
    if board[0][n - 1] != 0:
        player = board[0][n - 1]
        won = True
        for i in range(n):
            if board[i][n - 1 - i] != player:
                won = False
        if won:
            return player

    # check for empty spaces
    for i in range(n):
        for j in range(n):
            if board[i][j] == 0:
                return -1

    # no empty spaces means draw
    return 0

def make_move(board, player, row, col):
    if board[row][col] == 0:
        board[row][col] = player
        return True
    else:
        return False


def play_move(board, player, location):
    col = ord(location[0]) - ord("A")
    row = int(location[1]) - 1
    result = make_move(board, player, row, col)
    if result == True:
        draw_board(board)
    return result

def get_move(board, player):
    valid = False
    while valid == False:
        location = input("Player " + str(player) + ", enter a location: ")
        valid = play_move(board, player, location)
    return True

def generate_moves(board, player, original_player, depth):
    moves = []
    scores = []

    # stop searching if we reach the maximum depth
    # or the game is over
    if depth == 0 or check_game(board) != -1:
        score = get_score(board, original_player)
        return [()], [score]
    n = len(board)

    # look at every possible move
    for i in range(n):
        for j in range(n):
            if board[i][j] == 0:

                # make a copy of the board
                new_board = []

                for row in board:
                    new_board.append(row.copy())
                    
                # make the move on the copied board
                new_board[i][j] = player

                # search the next level
                next_moves, next_scores = generate_moves(
                    new_board,
                    3 - player,
                    original_player,
                    depth - 1
                )

                # add this move to the beginning
                # of each move sequence
                for k in range(len(next_moves)):
                    moves.append(
                        ((player, (i, j), depth),) + next_moves[k]
                    )

                    scores.append(next_scores[k])

    return moves, scores


def get_score(board, player):
    result = check_game(board)

    # computer/player wins
    if result == player:
        return 1

    # opponent wins
    elif result == 3 - player:
        return -1

    # draw
    elif result == 0:
        return 0

    # game is still going
    else:
        return 0


def tree_search(t, depth=1):
    # combine the best and worst outcomes,
    # while discounting outcomes farther away

    if isinstance(t[0], list):
        return sum(
            [tree_search(item, depth + 1) / depth for item in t]
        )
    else:
        return max(t) + min(t)


def pick_move(board, player, depth=5):
    moves, scores = generate_moves(
        board,
        player,
        player,
        depth
    )

    result = []
    for score in scores:
        result.append(score)

    # find the move with the highest score
    move_index = result.index(max(result))

    # return the first move in that sequence
    return moves[move_index][0]


def computer_move(board, depth=5):
    # player 2 is the computer
    move = pick_move(board, 2, depth)

    row = move[1][0]
    col = move[1][1]

    board[row][col] = 2


def start_game(n, depth=5):
    board = create_board(n)

    draw_board(board)

    game_status = -1
    player = 1

    while game_status == -1:

        # player 1 is the human
        if player == 1:
            get_move(board, player)

        # player 2 is the computer
        else:
            computer_move(board, depth)
            draw_board(board)

        # check if the game is over
        game_status = check_game(board)

        # switch players
        if game_status == -1:
            if player == 1:
                player = 2
            else:
                player = 1

    # print the result
    if game_status == 1:
        print("Player 1 wins!")

    elif game_status == 2:
        print("Computer wins!")

    else:
        print("The game is a draw!")
````

````python
def create_board(n):
    board = []

    for i in range(n):
        row = []

        for j in range(n):
            row.append(0)

        board.append(row)

    return board


def draw_board(board):
    n = len(board)

    print("    ", end="")

    for j in range(n):
        print(chr(65 + j) + "   ", end="")

    print()

    for i in range(n):
        print("  " + "--- " * n)

        print(str(i + 1) + " ", end="")

        for j in range(n):
            if board[i][j] == 0:
                print("|   ", end="")
            elif board[i][j] == 1:
                print("| X ", end="")
            elif board[i][j] == 2:
                print("| O ", end="")

        print("|")

    print("  " + "--- " * n)


def check_game(board):
    n = len(board)

    # check rows
    for i in range(n):
        if board[i][0] != 0:
            player = board[i][0]
            won = True

            for j in range(n):
                if board[i][j] != player:
                    won = False

            if won:
                return player

    # check columns
    for j in range(n):
        if board[0][j] != 0:
            player = board[0][j]
            won = True

            for i in range(n):
                if board[i][j] != player:
                    won = False

            if won:
                return player

    # check main diagonal
    if board[0][0] != 0:
        player = board[0][0]
        won = True

        for i in range(n):
            if board[i][i] != player:
                won = False

        if won:
            return player

    # check other diagonal
    if board[0][n - 1] != 0:
        player = board[0][n - 1]
        won = True

        for i in range(n):
            if board[i][n - 1 - i] != player:
                won = False

        if won:
            return player

    # check for empty spaces
    for i in range(n):
        for j in range(n):
            if board[i][j] == 0:
                return -1

    return 0


def make_move(board, player, row, col):
    if board[row][col] == 0:
        board[row][col] = player
        return True
    else:
        return False


def play_move(board, player, location):
    col = ord(location[0]) - ord("A")
    row = int(location[1]) - 1

    result = make_move(board, player, row, col)

    if result == True:
        draw_board(board)

    return result


def get_move(board, player):
    valid = False

    while valid == False:
        location = input("Player " + str(player) + ", enter a location: ")

        valid = play_move(board, player, location)

    return True


def get_score(board, player):
    result = check_game(board)

    if result == player:
        return 1

    elif result == 3 - player:
        return -1

    else:
        return 0


def generate_moves(board, player, original_player, depth):
    moves = []
    scores = []

    # stop searching at maximum depth
    # or when the game is over
    if depth == 0 or check_game(board) != -1:
        score = get_score(board, original_player)

        return [()], [score]

    n = len(board)

    for i in range(n):
        for j in range(n):

            if board[i][j] == 0:

                # make a copy of the board
                new_board = []

                for row in board:
                    new_board.append(row.copy())

                # make the move
                new_board[i][j] = player

                # search the next possible moves
                next_moves, next_scores = generate_moves(
                    new_board,
                    3 - player,
                    original_player,
                    depth - 1
                )

                # add the current move to each move sequence
                for k in range(len(next_moves)):
                    moves.append(
                        ((player, (i, j), depth),) + next_moves[k]
                    )

                    scores.append(next_scores[k])

    return moves, scores


def tree_search(t, depth=1):
    if isinstance(t[0], list):
        return sum(
            [tree_search(item, depth + 1) / depth for item in t]
        )
    else:
        return max(t) + min(t)


def pick_move(board, player, depth=5):
    moves, scores = generate_moves(
        board,
        player,
        player,
        depth
    )

    result = []

    for score in scores:
        result.append(score)

    move_index = result.index(max(result))

    return moves[move_index][0]


def computer_move(board, player, depth):
    move = pick_move(board, player, depth)

    row = move[1][0]
    col = move[1][1]

    board[row][col] = player


def computer_game(n, smart_depth, other_depth):
    board = create_board(n)

    player = 1
    game_status = -1

    while game_status == -1:

        if player == 1:
            computer_move(board, 1, smart_depth)

        else:
            computer_move(board, 2, other_depth)

        game_status = check_game(board)

        if game_status == -1:
            if player == 1:
                player = 2
            else:
                player = 1

    return game_status


def play_games(n, smart_depth, other_depth):
    player_1_wins = 0
    player_2_wins = 0
    draws = 0

    for i in range(10):

        result = computer_game(
            n,
            smart_depth,
            other_depth
        )

        if result == 1:
            player_1_wins += 1

        elif result == 2:
            player_2_wins += 1

        else:
            draws += 1

    win_rate = player_1_wins / 10 * 100

    print()
    print(n, "x", n, "grid")
    print("----------------")
    print("Player 1 wins:", player_1_wins)
    print("Player 2 wins:", player_2_wins)
    print("Draws:", draws)
    print("Smarter player's win rate:", win_rate, "%")
````

````python
print("Testing 3x3 board")
play_games(3, 5, 2)

print()
print("Testing 4x4 board")
play_games(4, 3, 1)

print()
print("Testing 5x5 board")
play_games(5, 2, 1)
````
