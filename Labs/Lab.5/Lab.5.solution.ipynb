```python
class matrix:
    def __init__(self, n, m=None):
        # Create a matrix filled with zeros
        if isinstance(n, int) and isinstance(m, int):
            if n < 0 or m < 0:
                raise ValueError("Matrix dimensions cannot be negative")

            self.data = []
            for i in range(n):
                row = []
                for j in range(m):
                    row.append(0)
                self.data.append(row)

        # Create a matrix from a list of lists
        elif isinstance(n, list) and m is None:
            if not all(isinstance(row, list) for row in n):
                raise ValueError("Matrix must be a list of lists")

            if len(n) > 0:
                columns = len(n[0])

                for row in n:
                    if len(row) != columns:
                        raise ValueError("All rows must have the same number of columns")

            self.data = [row[:] for row in n]

        else:
            raise ValueError("Provide dimensions or a list of lists")

    # Get an element or a slice
    def __getitem__(self, key):
        if isinstance(key, tuple):
            i, j = key

            # Access one element
            if isinstance(i, int) and isinstance(j, int):
                return self.data[i][j]

            # Select rows and columns
            rows = self.data[i]

            if isinstance(i, int):
                if isinstance(j, slice):
                    return rows[j]
                return rows[j]

            if isinstance(j, slice):
                new_data = []
                for row in rows:
                    new_data.append(row[j])
                return matrix(new_data)

            new_data = []
            for row in rows:
                new_data.append([row[j]])
            return matrix(new_data)

        # Select a single row
        if isinstance(key, int):
            return self.data[key]

        # Select multiple rows
        if isinstance(key, slice):
            return matrix(self.data[key])

    # Assign an element, row, or entire matrix
    def __setitem__(self, key, value):
        if isinstance(key, tuple):
            i, j = key
            self.data[i][j] = value

        elif isinstance(key, slice):
            if isinstance(value, matrix):
                new_data = value.data
            elif isinstance(value, list):
                new_data = value
            else:
                raise ValueError("Value must be a matrix or list of lists")

            if len(new_data) != len(self.data):
                raise ValueError("Matrices must have the same size")

            for i in range(len(self.data)):
                if len(new_data[i]) != len(self.data[i]):
                    raise ValueError("Matrices must have the same size")

            self.data = [row[:] for row in new_data]

        else:
            if len(value) != len(self.data[key]):
                raise ValueError("Row must have the correct number of columns")

            self.data[key] = list(value)

    # Return the shape of the matrix
    def shape(self):
        n = len(self.data)

        if n == 0:
            return (0, 0)

        m = len(self.data[0])
        return (n, m)

    # Return the transpose
    def transpose(self):
        n, m = self.shape()
        result = []

        for j in range(m):
            row = []
            for i in range(n):
                row.append(self.data[i][j])
            result.append(row)

        return matrix(result)

    # Return the nth row as a new matrix
    def row(self, n):
        return matrix([self.data[n][:]])

    # Return the nth column as a new matrix
    def column(self, n):
        result = []

        for i in range(len(self.data)):
            result.append([self.data[i][n]])

        return matrix(result)

    # Return a list of lists
    def to_list(self):
        return [row[:] for row in self.data]

    # Return a smaller block
    # n0:n1 selects columns; m0:m1 selects rows
    def block(self, n_0, n_1, m_0, m_1):
        result = []

        for i in range(m_0, m_1):
            result.append(self.data[i][n_0:n_1])

        return matrix(result)

    # Display the matrix
    def __str__(self):
        return str(self.data)
```

3. Write functions that create special matrices (note these are standalone functions, not member functions of your `matrix` class):
    * `constant(n,m,c)`: returns a `n` by `m` matrix filled with floats of value `c`.
    * `zeros(n,m)` and `ones(n,m)`: return `n` by `m` matrices filled with floats of value `0` and `1`, respectively.
    * `eye(n)`: returns the n by n identity matrix.

```python
def constant(n, m, c):
    result = []

    for i in range(n):
        row = []
        for j in range(m):
            row.append(float(c))
        result.append(row)

    return matrix(result)


def zeros(n, m):
    return constant(n, m, 0.0)


def ones(n, m):
    return constant(n, m, 1.0)


def eye(n):
    result = []

    for i in range(n):
        row = []

        for j in range(n):
            if i == j:
                row.append(1.0)
            else:
                row.append(0.0)

        result.append(row)

    return matrix(result)
```

6. Demonstrate the basic properties of matrices with your matrix class by creating two 2 by 2 example matrices using your Matrix class and illustrating the following:

$$
(AB)C=A(BC)
$$
$$
A(B+C)=AB+AC
$$
$$
AB\neq BA
$$
$$
AI=A
$$

```python
A = matrix([[1, 2],
            [3, 4]])

B = matrix([[2, 0],
            [1, 2]])

C = matrix([[1, 1],
            [0, 2]])

# Create the 2 by 2 identity matrix
I = eye(2)


# 1. Associative property: (AB)C = A(BC)
left = A.mat_mult(B).mat_mult(C)
right = A.mat_mult(B.mat_mult(C))

print("1. Associative property")
print("(AB)C =", left)
print("A(BC) =", right)
print("Equal:", left.equals(right))


# 2. Distributive property: A(B + C) = AB + AC
left = A.mat_mult(B.add(C))
right = A.mat_mult(B).add(A.mat_mult(C))

print("\n2. Distributive property")
print("A(B + C) =", left)
print("AB + AC =", right)
print("Equal:", left.equals(right))


# 3. Matrix multiplication is not commutative: AB != BA
AB = A.mat_mult(B)
BA = B.mat_mult(A)

print("\n3. Non-commutative property")
print("AB =", AB)
print("BA =", BA)
print("AB != BA:", not AB.equals(BA))


# 4. Identity property: AI = A
result = A.mat_mult(I)

print("\n4. Identity property")
print("AI =", result)
print("A =", A)
print("Equal:", result.equals(A))
```
